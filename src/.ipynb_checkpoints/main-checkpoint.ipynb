```python
import cv2
import numpy as np
import os
import json
from pathlib import Path
import matplotlib.pyplot as plt
from typing import Dict, List, Tuple, Optional

class BEVConverter:
    def __init__(self, dataset_path: str, camera_config_path: Optional[str] = None):
        """
        Initialize BEV converter with dataset path and camera configuration
        
        Args:
            dataset_path: Path to the dataset root (e.g., 'D:/Mtech/Disseration/BEV_Adas/Dataset/images0')
            camera_config_path: Path to camera configuration file (optional)
        """
        self.dataset_path = Path(dataset_path)
        self.rgb_path = self.dataset_path / 'train' / 'rgb'
        self.camera_config_path = camera_config_path or self.dataset_path / 'train' / 'cameraconfig'
        
        # Camera positions relative to vehicle center (in meters)
        self.camera_positions = {
            'front': {'x': 1.5, 'y': 0, 'z': 1.2, 'yaw': 0},
            'rear': {'x': -1.5, 'y': 0, 'z': 1.2, 'yaw': 180},
            'left': {'x': 0, 'y': 1.0, 'z': 1.2, 'yaw': 90},
            'right': {'x': 0, 'y': -1.0, 'z': 1.2, 'yaw': -90}
        }
        
        # Default camera intrinsics (adjust based on your camera config)
        self.default_intrinsics = {
            'fx': 800, 'fy': 800,  # focal lengths
            'cx': 640, 'cy': 360,  # principal point
            'k1': 0, 'k2': 0, 'k3': 0,  # radial distortion
            'p1': 0, 'p2': 0  # tangential distortion
        }
        
        # BEV parameters
        self.bev_width = 800
        self.bev_height = 800
        self.meters_per_pixel = 0.1  # 10cm per pixel
        self.bev_range_x = (-20, 20)  # meters
        self.bev_range_y = (-20, 20)  # meters
        
        self.load_camera_config()
        
    def load_camera_config(self):
        """Load camera configuration if available"""
        config_file = self.camera_config_path / 'camera_config.json'
        if config_file.exists():
            with open(config_file, 'r') as f:
                self.camera_config = json.load(f)
                print(f"Loaded camera configuration from {config_file}")
        else:
            print("No camera config found, using default parameters")
            self.camera_config = {}
    
    def get_camera_intrinsics(self, camera_name: str) -> Dict:
        """Get camera intrinsics for specific camera"""
        if camera_name in self.camera_config:
            return self.camera_config[camera_name].get('intrinsics', self.default_intrinsics)
        return self.default_intrinsics
    
    def get_perspective_transform_points(self, camera_name: str, img_height: int, img_width: int):
        """Get source and destination points for perspective transformation"""
        
        # Define region of interest based on camera
        if camera_name == 'front':
            # Front camera - trapezoid focusing on road ahead
            src_points = np.array([
                [img_width * 0.1, img_height * 0.9],    # bottom-left
                [img_width * 0.9, img_height * 0.9],    # bottom-right  
                [img_width * 0.6, img_height * 0.4],    # top-right (vanishing point)
                [img_width * 0.4, img_height * 0.4]     # top-left (vanishing point)
            ], dtype=np.float32)
            
            # BEV destination - front area
            dst_points = np.array([
                [self.bev_width * 0.3, self.bev_height * 0.8],   # bottom-left
                [self.bev_width * 0.7, self.bev_height * 0.8],   # bottom-right
                [self.bev_width * 0.7, self.bev_height * 0.2],   # top-right
                [self.bev_width * 0.3, self.bev_height * 0.2]    # top-left
            ], dtype=np.float32)
            
        elif camera_name == 'rear':
            src_points = np.array([
                [img_width * 0.1, img_height * 0.9],
                [img_width * 0.9, img_height * 0.9],
                [img_width * 0.6, img_height * 0.4],
                [img_width * 0.4, img_height * 0.4]
            ], dtype=np.float32)
            
            dst_points = np.array([
                [self.bev_width * 0.3, self.bev_height * 0.2],
                [self.bev_width * 0.7, self.bev_height * 0.2],
                [self.bev_width * 0.7, self.bev_height * 0.8],
                [self.bev_width * 0.3, self.bev_height * 0.8]
            ], dtype=np.float32)
            
        elif camera_name == 'left':
            src_points = np.array([
                [img_width * 0.1, img_height * 0.9],
                [img_width * 0.9, img_height * 0.9],
                [img_width * 0.6, img_height * 0.4],
                [img_width * 0.4, img_height * 0.4]
            ], dtype=np.float32)
            
            dst_points = np.array([
                [self.bev_width * 0.2, self.bev_height * 0.7],
                [self.bev_width * 0.8, self.bev_height * 0.7],
                [self.bev_width * 0.8, self.bev_height * 0.3],
                [self.bev_width * 0.2, self.bev_height * 0.3]
            ], dtype=np.float32)
            
        else:  # right camera
            src_points = np.array([
                [img_width * 0.1, img_height * 0.9],
                [img_width * 0.9, img_height * 0.9],
                [img_width * 0.6, img_height * 0.4],
                [img_width * 0.4, img_height * 0.4]
            ], dtype=np.float32)
            
            dst_points = np.array([
                [self.bev_width * 0.2, self.bev_height * 0.3],
                [self.bev_width * 0.8, self.bev_height * 0.3],
                [self.bev_width * 0.8, self.bev_height * 0.7],
                [self.bev_width * 0.2, self.bev_height * 0.7]
            ], dtype=np.float32)
        
        return src_points, dst_points
    
    def world_to_bev(self, world_x: float, world_y: float) -> Tuple[int, int]:
        """Convert world coordinates to BEV pixel coordinates"""
        # Convert world coordinates to BEV image coordinates
        bev_x = int((world_x - self.bev_range_x[0]) / self.meters_per_pixel)
        bev_y = int((self.bev_range_y[1] - world_y) / self.meters_per_pixel)
        
        return bev_x, bev_y
    
    def create_homography_matrix(self, camera_name: str, img_height: int, img_width: int) -> np.ndarray:
        """Create homography matrix for perspective transformation"""
        src_points, dst_points = self.get_perspective_transform_points(camera_name, img_height, img_width)
        homography = cv2.getPerspectiveTransform(src_points, dst_points)
        return homography
    
    def convert_image_to_bev(self, image: np.ndarray, camera_name: str) -> np.ndarray:
        """Convert single camera image to BEV"""
        img_height, img_width = image.shape[:2]
        homography = self.create_homography_matrix(camera_name, img_height, img_width)
        
        # Apply perspective transformation
        bev_img = cv2.warpPerspective(
            image, homography, 
            (self.bev_width, self.bev_height),
            flags=cv2.INTER_LINEAR,
            borderMode=cv2.BORDER_CONSTANT,
            borderValue=(0, 0, 0)
        )
        
        return bev_img
    
    def create_roi_mask(self, camera_name: str) -> np.ndarray:
        """Create ROI mask for each camera to avoid overlapping regions"""
        mask = np.zeros((self.bev_height, self.bev_width), dtype=np.uint8)
        
        if camera_name == 'front':
            # Front camera covers front area (trapezoid)
            points = np.array([
                [self.bev_width * 0.2, self.bev_height * 0.5],  # center-left
                [self.bev_width * 0.8, self.bev_height * 0.5],  # center-right
                [self.bev_width * 0.9, 0],                      # top-right
                [self.bev_width * 0.1, 0]                       # top-left
            ], dtype=np.int32)
            
        elif camera_name == 'rear':
            # Rear camera covers rear area
            points = np.array([
                [self.bev_width * 0.1, self.bev_height],        # bottom-left
                [self.bev_width * 0.9, self.bev_height],        # bottom-right
                [self.bev_width * 0.8, self.bev_height * 0.5],  # center-right
                [self.bev_width * 0.2, self.bev_height * 0.5]   # center-left
            ], dtype=np.int32)
            
        elif camera_name == 'left':
            # Left camera covers left area
            points = np.array([
                [0, self.bev_height * 0.1],                     # top-left
                [self.bev_width * 0.5, self.bev_height * 0.2],  # top-center
                [self.bev_width * 0.5, self.bev_height * 0.8],  # bottom-center
                [0, self.bev_height * 0.9]                      # bottom-left
            ], dtype=np.int32)
            
        else:  # right camera
            # Right camera covers right area
            points = np.array([
                [self.bev_width, self.bev_height * 0.1],        # top-right
                [self.bev_width * 0.5, self.bev_height * 0.2],  # top-center
                [self.bev_width * 0.5, self.bev_height * 0.8],  # bottom-center
                [self.bev_width, self.bev_height * 0.9]         # bottom-right
            ], dtype=np.int32)
        
        cv2.fillPoly(mask, [points], 255)
        return mask
    
    def fuse_bev_images(self, bev_images: Dict[str, np.ndarray]) -> np.ndarray:
        """Fuse multiple BEV images into single unified BEV with alpha blending"""
        fused_bev = np.zeros((self.bev_height, self.bev_width, 3), dtype=np.float32)
        weight_sum = np.zeros((self.bev_height, self.bev_width), dtype=np.float32)
        
        for camera_name, bev_img in bev_images.items():
            if bev_img is not None:
                # Create ROI mask for this camera
                mask = self.create_roi_mask(camera_name)
                
                # Convert mask to 3-channel
                mask_3ch = cv2.merge([mask, mask, mask]) / 255.0
                
                # Apply weighted blending
                weighted_img = bev_img.astype(np.float32) * mask_3ch
                fused_bev += weighted_img
                weight_sum += mask / 255.0
        
        # Normalize by weights to avoid over-bright areas
        weight_sum_3ch = np.stack([weight_sum, weight_sum, weight_sum], axis=2)
        weight_sum_3ch[weight_sum_3ch == 0] = 1  # Avoid division by zero
        
        fused_bev = fused_bev / weight_sum_3ch
        fused_bev = np.clip(fused_bev, 0, 255).astype(np.uint8)
        
        return fused_bev
    
    def process_frame(self, frame_id: str, verbose: bool = False) -> np.ndarray:
        """Process all camera images for a single frame"""
        bev_images = {}
        found_cameras = []
        
        for camera_name in ['front', 'rear', 'left', 'right']:
            # Try multiple image extensions and paths
            possible_paths = [
                self.rgb_path / camera_name / f"{frame_id}.jpg",
                self.rgb_path / camera_name / f"{frame_id}.jpeg",
                self.rgb_path / camera_name / f"{frame_id}.png",
                self.rgb_path / camera_name / f"{frame_id}.JPG",
                self.rgb_path / camera_name / f"{frame_id}.JPEG",
                self.rgb_path / camera_name / f"{frame_id}.PNG"
            ]
            
            img_path = None
            for path in possible_paths:
                if path.exists():
                    img_path = path
                    break
            
            if img_path:
                # Load image
                img = cv2.imread(str(img_path))
                if img is not None:
                    # Convert to BEV
                    bev_img = self.convert_image_to_bev(img, camera_name)
                    bev_images[camera_name] = bev_img
                    found_cameras.append(camera_name)
                    if verbose:
                        print(f"✓ Processed {camera_name} camera")
                elif verbose:
                    print(f"✗ Failed to load {camera_name} image")
            elif verbose:
                print(f"✗ {camera_name} image not found")
        
        if verbose:
            print(f"Found cameras: {', '.join(found_cameras)}")
        
        # Fuse all BEV images
        if bev_images:
            fused_bev = self.fuse_bev_images(bev_images)
            return fused_bev
        else:
            if verbose:
                print(f"No valid images found for frame {frame_id}")
            return None
    
    def process_dataset(self, output_dir: str, max_frames: Optional[int] = None, verbose: bool = False):
        """Process entire dataset and save BEV images"""
        output_path = Path(output_dir)
        output_path.mkdir(parents=True, exist_ok=True)
        
        # Try to find frame IDs from any available camera folder
        frame_ids = []
        for camera_name in ['front', 'rear', 'left', 'right']:
            camera_dir = self.rgb_path / camera_name
            if camera_dir.exists():
                if verbose:
                    print(f"Checking {camera_name} directory...")
                # Look for all image extensions
                frame_files = []
                for ext in ['*.jpg', '*.jpeg', '*.png', '*.JPG', '*.JPEG', '*.PNG']:
                    frame_files.extend(list(camera_dir.glob(ext)))
                
                if frame_files:
                    frame_ids = [f.stem for f in frame_files]
                    print(f"Found {len(frame_ids)} frames in {camera_name} directory")
                    break
        
        if not frame_ids:
            print("❌ No frames found in any camera directory")
            return
        
        if max_frames:
            frame_ids = frame_ids[:max_frames]
        
        print(f"🚀 Processing {len(frame_ids)} frames...")
        success_count = 0
        
        for i, frame_id in enumerate(frame_ids):
            if verbose or (i + 1) % 10 == 0:
                print(f"Processing frame {i+1}/{len(frame_ids)}: {frame_id}")
            
            # Process frame
            bev_result = self.process_frame(frame_id, verbose=verbose)
            
            if bev_result is not None:
                # Save BEV image
                output_file = output_path / f"bev_{frame_id}.jpg"
                success = cv2.imwrite(str(output_file), bev_result)
                if success:
                    success_count += 1
                    if verbose:
                        print(f"✅ Saved: {output_file}")
                else:
                    print(f"❌ Failed to save: {output_file}")
            
            # Progress indicator
            if (i + 1) % 10 == 0:
                print(f"Progress: {i+1}/{len(frame_ids)} ({success_count} successful)")
        
        print(f"🎉 Completed! {success_count}/{len(frame_ids)} frames processed successfully")
    
    def visualize_frame(self, frame_id: str, save_path: Optional[str] = None):
        """Visualize original images and resulting BEV for a single frame"""
        fig, axes = plt.subplots(2, 3, figsize=(15, 10))
        fig.suptitle(f'BEV Conversion - Frame {frame_id}', fontsize=16)
        
        # Load and display original images
        cameras = ['front', 'left', 'right', 'rear']
        positions = [(0, 1), (1, 0), (1, 2), (1, 1)]  # subplot positions
        
        for camera, pos in zip(cameras, positions):
            # Try multiple extensions
            img_path = None
            for ext in ['.jpg', '.jpeg', '.png', '.JPG', '.JPEG', '.PNG']:
                test_path = self.rgb_path / camera / f"{frame_id}{ext}"
                if test_path.exists():
                    img_path = test_path
                    break
            
            if img_path and img_path.exists():
                img = cv2.imread(str(img_path))
                if img is not None:
                    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
                    axes[pos].imshow(img_rgb)
                    axes[pos].set_title(f'{camera.capitalize()} Camera')
                    axes[pos].axis('off')
                else:
                    axes[pos].text(0.5, 0.5, f'{camera.capitalize()}\nFailed to load', 
                                 ha='center', va='center', transform=axes[pos].transAxes)
                    axes[pos].axis('off')
            else:
                axes[pos].text(0.5, 0.5, f'{camera.capitalize()}\nNot found', 
                             ha='center', va='center', transform=axes[pos].transAxes)
                axes[pos].axis('off')
        
        # Generate and display BEV
        bev_result = self.process_frame(frame_id, verbose=True)
        if bev_result is not None:
            bev_rgb = cv2.cvtColor(bev_result, cv2.COLOR_BGR2RGB)
            axes[0, 0].imshow(bev_rgb)
            axes[0, 0].set_title('Fused BEV')
            axes[0, 0].axis('off')
        else:
            axes[0, 0].text(0.5, 0.5, 'BEV\nGeneration Failed', 
                          ha='center', va='center', transform=axes[0, 0].transAxes)
            axes[0, 0].axis('off')
        
        # Remove empty subplot
        axes[0, 2].axis('off')
        
        plt.tight_layout()
        
        if save_path:
            plt.savefig(save_path, dpi=300, bbox_inches='tight')
            print(f"Visualization saved: {save_path}")
        
        plt.show()


def check_directory_structure(dataset_path: str):
    """Check and display the actual directory structure"""
    base_path = Path(dataset_path)
    print(f"📁 Checking directory structure at: {base_path}")
    print(f"Base path exists: {base_path.exists()}")
    
    if base_path.exists():
        print("\n📂 Directory contents:")
        dirs_found = []
        images_found = 0
        
        for item in base_path.rglob("*"):
            if item.is_dir():
                dirs_found.append(str(item.relative_to(base_path)))
            elif item.suffix.lower() in ['.jpg', '.jpeg', '.png']:
                images_found += 1
        
        # Show directory structure
        for dir_path in sorted(dirs_found)[:20]:  # Limit to first 20 dirs
            print(f"  📁 {dir_path}")
        
        if len(dirs_found) > 20:
            print(f"  ... and {len(dirs_found) - 20} more directories")
        
        print(f"\n📸 Total images found: {images_found}")
    else:
        print("❌ Path does not exist!")


def main():
    """Main function to demonstrate BEV conversion"""
    # Use proper Windows path
    dataset_path = r"D:\Mtech\Disseration\BEV_Adas\Dataset\images0"
    
    # First check the directory structure
    check_directory_structure(dataset_path)
    
    # Initialize converter
    converter = BEVConverter(dataset_path)
    
    # Check multiple possible locations for images
    possible_paths = [
        converter.rgb_path / 'front',  # train/rgb/front
        Path(dataset_path) / 'rgb' / 'front',  # images0/rgb/front
        Path(dataset_path) / 'front',  # images0/front
        Path(dataset_path) / 'train' / 'front'  # train/front
    ]
    
    front_dir = None
    for path in possible_paths:
        print(f"Checking path: {path}")
        if path.exists():
            front_dir = path
            print(f"Found front directory at: {front_dir}")
            break
    
    if front_dir:
        # Look for common image extensions
        frame_files = []
        for ext in ['*.jpg', '*.jpeg', '*.png', '*.JPG', '*.JPEG', '*.PNG']:
            frame_files.extend(list(front_dir.glob(ext)))
        
        print(f"Found {len(frame_files)} image files")
        
        if frame_files:
            test_frame_id = frame_files[0].stem
            print(f"Testing with frame: {test_frame_id}")
            
            # Update converter paths if needed
            if front_dir != converter.rgb_path / 'front':
                converter.rgb_path = front_dir.parent
                print(f"Updated RGB path to: {converter.rgb_path}")
            
            # Visualize single frame
            print(f"\n🎨 Creating visualization for frame: {test_frame_id}")
            converter.visualize_frame(test_frame_id, f"bev_visualization_{test_frame_id}.png")
            
            # Process small batch
            print(f"\n🔄 Processing batch of frames...")
            output_dir = "bev_output"
            converter.process_dataset(output_dir, max_frames=5, verbose=False)
            
        else:
            print("No image files found in any front camera directory")
            print("Supported formats: JPG, JPEG, PNG")
    else:
        print("Front camera directory not found in any expected location")
        print("Expected locations:")
        for path in possible_paths:
            print(f"  - {path}")


if __name__ == "__main__":
    main()
```
